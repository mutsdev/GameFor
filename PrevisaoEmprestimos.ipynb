```python
import pandas as pd

data = pd.read_csv("loan_approval_dataset.csv")

list(data.columns)
```

```python
data.head()
```

```python
data.columns = data.columns.str.strip()

data.rename(columns={
    'no_of_dependents': 'Numero Dependentes',
    'education': 'Graduado',
    'self_employed': 'Autônomo',
    'income_annum': 'Renda Anual Dolares',
    'loan_amount': 'Valor Emprestimo Dolares',
    'loan_term': 'Prazo do Emprestimo',
    'cibil_score': 'Pontuacao de Credito',
    'residential_assets_value': 'Valores dos Ativos Residenciais Dolares',
    'commercial_assets_value': 'Valor de Ativos Comerciais Dolares',
    'luxury_assets_value': 'Valores de Ativos de Luxo Dolares',
    'bank_asset_value': 'Valor de Ativos Bancarios Dolares',
    'loan_status': 'Status do Emprestimo'
}, inplace=True)

list(data.columns)
```

```python
data.head()
```

```python
data['Graduado'] = data['Graduado'].str.strip()
data['Graduado'] = data['Graduado'].replace({'Graduate': 1, 'Not Graduate': 0})
data.head()
```

```python
data['Autônomo'] = data['Autônomo'].str.strip()
data['Autônomo'] = data['Autônomo'].replace({'Yes': 1, 'No': 0})
data['Autônomo'].unique()
```

```python
data['Status do Emprestimo'] = data['Status do Emprestimo'].str.strip()
data['Status do Emprestimo'] = data['Status do Emprestimo'].replace({'Approved': 1, 'Rejected': 0})
data['Status do Emprestimo'].unique()
```

```python
import numpy as np
from sklearn.model_selection import train_test_split

data.columns = data.columns.str.strip()

X = data[['Numero Dependentes', 'Graduado', 'Autônomo', 'Renda Anual Dolares',
          'Valor Emprestimo Dolares', 'Prazo do Emprestimo', 'Pontuacao de Credito',
          'Valores dos Ativos Residenciais Dolares', 'Valor de Ativos Comerciais Dolares',
          'Valores de Ativos de Luxo Dolares', 'Valor de Ativos Bancarios Dolares']]

Y = data['Status do Emprestimo']

X = X.to_numpy()
Y = Y.to_numpy()

SEED = 432849

x_train, x_test, y_train, y_test = train_test_split(X, Y, test_size=0.2, random_state=SEED, stratify=Y)

print(f"O treino será feito com {len(x_train)} elementos")
print(f"O teste será feito com {len(x_test)} elementos")
```

```python
from sklearn import tree
from sklearn.metrics import accuracy_score

model = tree.DecisionTreeClassifier(max_depth = 3, random_state = SEED)
model.fit(x_train, y_train)
predict = model.predict(x_test)
accuracy = accuracy_score(y_test, predict) * 100;

print(f"A acurácia foi de {accuracy:.2f}%")
```

```python
import graphviz
struct = tree.export_graphviz(model, feature_names = ['Numero Dependentes', 'Graduado', 'Autônomo', 'Renda Anual Dolares',
          'Valor Emprestimo Dolares', 'Prazo do Emprestimo', 'Pontuacao de Credito',
          'Valores dos Ativos Residenciais Dolares', 'Valor de Ativos Comerciais Dolares',
          'Valores de Ativos de Luxo Dolares', 'Valor de Ativos Bancarios Dolares'],
                              class_names = ['não', 'sim'], filled = True,fontname = "courier")
grafico = graphviz.Source(struct)
grafico
```
